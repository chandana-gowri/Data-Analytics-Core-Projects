# Task 6: Data Visualization – Python Charts for Insights

Retail Sales Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("sales_data_sample (1).csv", encoding="latin1")
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'])
df['MONTH'] = df['ORDERDATE'].dt.to_period('M').astype(str)
df.head()
```

## Bar Chart: Top 10 Product Lines by Sales

```python

top_products = df.groupby('PRODUCTLINE')['SALES'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_products.plot(kind='bar')
plt.title("Top 10 Product Lines by Sales")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

## Line Chart: Monthly Sales Trend

```python

monthly_sales = df.groupby('MONTH')['SALES'].sum()

plt.figure(figsize=(10,6))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

## Histogram: Sales Distribution

```python

plt.figure(figsize=(8,5))
plt.hist(df['SALES'], bins=20)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

## Scatter Plot: Sales vs Profit

```python
plt.figure(figsize=(8,5))
plt.scatter(df['SALES'], df['QUANTITYORDERED'])
plt.title("Sales vs Quantity Ordered")
plt.xlabel("Sales")
plt.ylabel("Quantity Ordered")
plt.grid(True)
plt.show()
```

```python
top_countries = df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=True).tail(10)

plt.figure(figsize=(10,6))
plt.barh(top_countries.index, top_countries.values, color='teal')
plt.title("Top 10 Countries by Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Country")
plt.grid(axis='x')
plt.show()
```

```python
deal_sales = df.groupby('DEALSIZE')['SALES'].sum()

plt.figure(figsize=(7,7))
plt.pie(
    deal_sales,
    labels=deal_sales.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['gold', 'lightskyblue', 'lightcoral']
)
plt.title("Sales Contribution by Deal Size")
plt.show()
```


## Key Insights
1. Sales are not evenly distributed; a few product categories contribute more to total revenue.

2. Some countries/regions show consistently higher sales compared to others.

3. Certain product lines have higher quantity orders, indicating strong demand.

4. Sales values show variation with presence of high-value outliers.

5. Visualization helps clearly identify top-performing products and regions.

6. The dataset shows patterns that can help improve marketing and inventory decisions.
